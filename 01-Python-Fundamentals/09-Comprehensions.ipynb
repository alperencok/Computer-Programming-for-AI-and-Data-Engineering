```python
# PYTHON COMPREHENSIONS
print("--- 1. LIST COMPREHENSION WITH 'IF' (FILTERING) ---")
# new_list = [ <EXPRESSION> for <ITEM> in <ITERABLE> if <CONDITION> ]

# Example: Filter scores greater than or equal to 50.
scores = [45, 80, 30, 95, 60]
# 'n' is the iteration variable representing each element in 'scores'.
passing_scores = [n for n in scores if n >= 50] 
print(f"Passing Scores: {passing_scores}")


print("\n--- 2. LIST COMPREHENSION WITH 'IF-ELSE' (TRANSFORMATION) ---")
# new_list = [ <TRUE_VALUE> if <CONDITION> else <FALSE_VALUE> for <ITEM> in <ITERABLE> ]

# Example: Map scores to Pass/Fail status.
statuses = ["Passed" if n >= 50 else "Failed" for n in scores]
# Evaluates condition first, then assigns value for each item.
print(f"Student Statuses: {statuses}")


print("\n--- 3. SINGLE LINE VARIABLE ASSIGNMENT (TERNARY OPERATOR) ---")
# <VARIABLE> = <TRUE_VALUE> if <CONDITION> else <FALSE_VALUE>

# Example: Check if system registration is active.
is_system_active = True
system_message = "Registration open." if is_system_active else "Registration closed."
print(f"System Message: {system_message}")


print("\n--- 4. ITERATING OVER LISTS INSIDE A DICTIONARY (NESTED COMPREHENSION) ---")
# A student may have multiple course scores stored in a list inside a dictionary.
student_grade_book = {
    "Alex": [50, 60, 45],
    "Alice": [90, 95, 100]
}

# [ <EXPRESSION> for <KEY>, <LIST> in <DICT>.items() for <ELEMENT> in <LIST> ]

# Example: Flatten all grades into a single list and add 5 bonus points to each.
all_bonus_grades = [grade + 5 for name, grade_list in student_grade_book.items() for grade in grade_list]
print(f"All Grades with Bonus: {all_bonus_grades}")


print("\n--- 5. ACCESSING TUPLES IN A DICTIONARY AND TYPE CASTING ---")

# Data from queries may arrive as string tuples.
# Example: "Course Title": ("Credits", "Capacity")
course_catalog = {
    "Mathematics": ("3", "40"),
    "Physics": ("4", "30"),
    "History": ("2", "50")
}

# [ <TYPE_CAST(VALUE)> for <KEY>, <TUPLE> in <DICT>.items() ]

# Example: Extract credit hours and convert them from string to integer: int()
numeric_credits = [int(info_tuple[0]) for course_name, info_tuple in course_catalog.items()]
print(f"Numeric Course Credits: {numeric_credits}")

# Filter: Course titles where capacity is greater than 35
large_courses = [course_name for course_name, info_tuple in course_catalog.items() if int(info_tuple[1]) > 35]
print(f"Courses with capacity > 35: {large_courses}")
```
