```python
# continue, break, pass
```

```python
numara_listesi=[10,20,30,40,50]
```

```python
print("Loop started")
for num in numara_listesi:
    print(num)
print("Loop ended")
```

```python
for num in numara_listesi:
    print(num)
    if num==40:
        break
```

```python
for num in numara_listesi:
    if num==40:
        continue
    print(num)
```

```python
if 4>3:
    pass
```

```python
for num in numara_listesi:
    pass
```
