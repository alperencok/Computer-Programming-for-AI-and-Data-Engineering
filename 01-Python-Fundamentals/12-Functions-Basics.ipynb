```python
# enumerate
```

```python
for x in range(len(ex_list)):
    print(ex_list[x], x)
```

```python
for element in enumerate(ex_list):
    print(element)
```

```python
for x, value in enumerate(ex_list):
    print(x, value)
```

```python
# list comprehension
```

```python
ex_list=[10,20,30,40,50]
```

```python
s_list=[]
for num in ex_list:
    s_list.append(num/2)
```

```python
s_list
```

```python
s_list=[num/2 for num in ex_list]
```

```python
s_list
```

```python
s_list=[]
for num in ex_list:
    if num % 6 == 0:
        s_list.append(num)
```

```python
s_list
```

```python
s_list=[num for num in ex_list if num % 6 == 0]
```

```python
s_list
```
