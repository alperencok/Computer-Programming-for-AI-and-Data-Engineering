```python
class Staff:
    def __init__ (self, _id, name,salary, workingYear, title):
        self._id=_id
        self.name=name
        self.salary=salary
        self.workingYear=workingYear
        self.title=title

    def _raise(self):
        self.salary= self.salary*1.1

    def info(self):
        print(self._id, self.name, self.salary, self.workingYear, self.title)
        
```

```python
class Engineer(Staff):
    def __init__ (self, _id, name,salary, workingYear, title):
        super().__init__(_id, name,salary, workingYear, title)

    def _raise(self):
        self.salary=self.salary*1.2
```

```python
class Worker(Staff):
    def __init__ (self, _id, name,salary, workingYear, title):
        super().__init__(_id, name,salary, workingYear, title)

    def _raise(self):
         self.salary=self.salary*1.15
```

```python
class Intern(Staff):
    def __init__ (self, _id, name,salary, workingYear, title):
        super().__init__(_id, name,salary, workingYear, title)

    def _raise(self):
         self.salary=self.salary
```

```python
class Factory:
    def __init__ (self, _id, name):
        self._id=_id
        self.name=name
        self.staffList=[]

    def staffAdd(self,staff):
        self.staffList.append(staff)

    def annualRaise(self):
        for staff in self.staffList:
            staff._raise()

    def staffCount(self):
        counter={"Engineer":0,
                 "Worker":0,
                 "Intern": 0}

        for staff in self.staffList:
            if isinstance(staff, Engineer):
                counter["Engineer"]+=1
            elif isinstance(staff, Worker):
                counter["Worker"]+=1
            elif isinstance(staff, Intern):
                counter["Intern"]+=1
            
        return counter

    def info(self):
        for staff in self.staffList:
            staff.info()
```

```python
e1= Engineer(100, "Ali", 90000, 3, "Software engineering")
e2= Engineer(101, "Ali", 90000, 3, "Software engineering")
w1= Worker(102, "Ayse", 50000, 3, "IT")
i1= Intern(103, "Bob", 20000, 3, "Intern")
i2= Intern(105, "Bob", 20000, 3, "Intern")
i3= Intern(106, "Bob", 20000, 3, "Intern")

f=Factory(1001, "ABC")
f.staffAdd(e1)
f.staffAdd(e2)
f.staffAdd(w1)
f.staffAdd(i1)
f.staffAdd(i2)
f.staffAdd(i3)
f.info()

f.annualRaise()
print("*********************")
f.info()
print("*********************")
print(f.staffCount())
```
