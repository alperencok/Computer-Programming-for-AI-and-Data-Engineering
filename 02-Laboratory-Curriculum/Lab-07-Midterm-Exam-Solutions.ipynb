## Based on the dictionary, write codes to answer the following questions

```python
stud_data = {
    "Kenan": {
        "scores": [85, 90, 78, 92],
        "credentials": ("CE1023", "std1@fsm.stud.edu.tr"),
        "address": {"city": "Istanbul", "street": "Yunus Emre Cd."}
    },
    "Arda": {
        "scores": [88, 85, 95, 89],
        "credentials": ("SE4056", "std2@fsm.stud.edu.tr"),
        "address": {"city": "Istanbul", "street": "Mevlevihane Cd."}
    },
    "Ismail": {
        "scores": [91, 84, 78, 88],
        "credentials": ("AIDE7089", "std3@fsm.stud.edu.tr"),
        "address": {"city": "Istanbul", "street": "Mevlana Cd."}
    }
}
```

```python
# a)	Using type conversion, change the third student’s first credential to: “AIDE0099”
temp_credentials = list(stud_data["Ismail"]["credentials"])  # Convert tuple to list
temp_credentials[0] = "AIDE0099"  # Change the first element
stud_data["Ismail"]["credentials"] = tuple(temp_credentials)  # Convert back to tuple
# Verify the change
print(stud_data["Ismail"]["credentials"])
```

```python
# b)	Using loop, change street names of all students into:  " Mevlevihane Cd.
for student, info in stud_data.items():
    info["address"]["street"] = "Mevlevihane Cd."
```

```python
print(stud_data['student_1']['scores'])
print(stud_data['student_2']['scores'])
```

```python
result = [(student, min([score for score in info["scores"] if score >= 78]))  for student, info in stud_data.items()]

for item in result:
    print(item)
```

```python
# c)	Using list comprehension, print students and their minimum score greater than 78
min_scores = [
    (student, min(data["scores"]))
    for student, data in stud_data.items()

]
for i in min_scores:
    print(i)
```

```python
# d) using dictionary comprehension, print names and average scores using dictionary comprehension
average_scores = {
    student: sum(data["scores"]) / len(data["scores"]) for student, data in stud_data.items()
}
print(average_scores)
```

```python
# e) write comprehension that combines all students' scores and print unique scores only.
unique_scores = { score  for student, data in stud_data.items()  if "scores" in data for score in data["scores"] }

print(unique_scores)
```
