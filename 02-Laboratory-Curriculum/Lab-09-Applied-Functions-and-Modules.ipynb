```python
students = [
    {"id": 1, "grades": {"math": 80, "cs": 90, "phy": 70}},
    {"id": 2, "grades": {"math": 60, "cs": 75, "phy": 85}},
    {"id": 3, "grades": {"math": 95, "cs": 100, "phy": 90}}
]
```

Requirements:
* Create a class named `Student`.
* Implement `__init__(self, student_id, grades)`:
  - `average_grade(self)`: Returns the average of all grades.
  - `best_course(self)`: Returns the course with the highest grade.
For all students:
* Print the student ID and average grade.


```python
class Student:
    def __init__(self,student_id,grades):
        
        self.student_id=student_id
        self.grades=grades
    def avg_grade(self):
        total=sum(self.grades.values())
        count=len(self.grades)
        avg=total/count
        
        return avg
    def best_course(self):
        return max(self.grades,key=self.grades.get)
    
student_obj=[]

for s in students:
    obj=Student(s["id"],s["grades"])
    student_obj.append(obj)
    
for s in student_obj:
    print(f"id: {s.student_id}, avg: {s.avg_grade()}, best: {s.best_course()}")
```

```python
products = {
    "apple": 10,
    "banana": 5,
    "milk": 20
}
cart_list = [
    {"id": 1, "items": {"apple": 2, "milk": 1}},
    {"id": 2, "items": {"banana": 5}},
]
```

Requirements:
* Create a class named `Cart`.
* Implement `__init__(self, cart_id, items)`:
  - `total_cost(self)`: Returns the total price of all items.
  - `most_expensive_product(self)`: Returns the item with highest unit price.
* Print total costs for all carts.


```python
class Card:
    def __init__(self,cart_id,items):
        self.cart_id=cart_id
        self.items=items
        
    def total_cost(self):
        total=0
        for item,qty in self.items.items():
            total+=products[item]*qty
        return total
    
    def most_exp(self):
        return max(self.items, key=products.get)
    
cart_objects = []

for c in cart_list:
    obj = Card(c["id"], c["items"])
    cart_objects.append(obj)
for cart in cart_objects:
    print(f"Cart ID: {cart.cart_id}, Total: {cart.total_cost()}, Most expensive item: {cart.most_exp()}")
```

```python
employees = [
    {"id": 1, "salary": 30000, "department": "IT"},
    {"id": 2, "salary": 25000, "department": "HR"},
    {"id": 3, "salary": 40000, "department": "IT"}
]
```

Requirements:
* Create a class named `Employee`.
* Implement `__init__(self, emp_id, salary, department)`:
  - `is_high_salary(self)`: Returns True if salary is above 30,000.
  - `increase_salary(self, percent)`: Increases salary by the given percentage.
* Print updated salaries for employees in the IT department.


```python
class Employee:
    def __init__(self, emp_id, salary, department):
        self.emp_id = emp_id
        self.salary = salary
        self.department = department
        
    def is_high_salary(self):
        return self.salary > 30000
    
    def increase_salary(self, percent):
        self.salary += self.salary * (percent / 100)
        
employee_objects = []

for e in employees:
    obj = Employee(e["id"], e["salary"], e["department"])
    employee_objects.append(obj)
for emp in employee_objects:
    if emp.department == "IT":
        emp.increase_salary(10)  # %10 zam
        print(f"ID: {emp.emp_id}, New Salary: {emp.salary}")
```

```python
import tkinter as tk
from tkinter import messagebox
import sqlite3

# -----------------------------
# DATABASE SETUP
# -----------------------------
def create_database():
    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute("""
    CREATE TABLE IF NOT EXISTS students (
        id INTEGER PRIMARY KEY AUTOINCREMENT,
        name TEXT,
        grade INTEGER
    )
    """)

    conn.commit()
    conn.close()

# -----------------------------
# DATA INSERTION
# -----------------------------
def add_student():
    name = entry_name.get()
    grade = entry_grade.get()

    if name == "" or grade == "":
        messagebox.showwarning("Warning", "Please fill in all fields!")
        return

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute("INSERT INTO students (name, grade) VALUES (?, ?)", (name, grade))

    conn.commit()
    conn.close()

    messagebox.showinfo("Success", "Student record added successfully!")

    entry_name.delete(0, tk.END)
    entry_grade.delete(0, tk.END)

# -----------------------------
# DISPLAY DATA IN GUI
# -----------------------------
def show_students():
    listbox.delete(0, tk.END)

    conn = sqlite3.connect("students.db")
    cursor = conn.cursor()

    cursor.execute("SELECT * FROM students")
    rows = cursor.fetchall()

    for row in rows:
        listbox.insert(tk.END, f"ID:{row[0]} | {row[1]} - {row[2]}")

    conn.close()

# -----------------------------
# MAIN GUI
# -----------------------------
create_database()

window = tk.Tk()
window.title("Student Management System")
window.geometry("350x400")

# Header
tk.Label(window, text="Student App", font=("Arial", 16)).pack(pady=10)

# Name
tk.Label(window, text="Name").pack()
entry_name = tk.Entry(window)
entry_name.pack()

# Grade
tk.Label(window, text="Grade").pack()
entry_grade = tk.Entry(window)
entry_grade.pack()

# Buttons
tk.Button(window, text="Add Student", command=add_student).pack(pady=10)
tk.Button(window, text="Show Students", command=show_students).pack()

# Listbox Area
listbox = tk.Listbox(window, width=40)
listbox.pack(pady=20)

window.mainloop()
```
