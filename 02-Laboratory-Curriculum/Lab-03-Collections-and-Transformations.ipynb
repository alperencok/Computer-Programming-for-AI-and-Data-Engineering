Take <sentence, keyword> from the user.

Convert the entire sentence to lowercase.

If the keyword appears in the sentence:

Replace the keyword with its uppercase version.

Otherwise:

> Keyword not found.

Print the result on the screen.

Enter a sentence: I am learning Python programming

Enter a keyword: python

Output:
i am learning PYTHON programming

````python
sentence=input("Enter a sentence:")
keyword=input("Enter a keyword:")

sentence=sentence.lower()
keyword=keyword.lower()

if keyword in sentence:
    sentence= sentence.replace(keyword, keyword.upper())
    print(sentence)
else:
    print("Keyword not found!")
````

Take a sentence from the user.

If the letter "e" appears in the sentence:

Print how many times the letter "e" occurs.

Otherwise:

Print "Letter not found".

````python
sentence=input("Enter a sentence")
sentence=sentence.lower()

if "e" in sentence:
    count_e=sentence.count("e")
    print(f"Letter e appears {count_e} times")
else:
    print("Letter e not found")
````
