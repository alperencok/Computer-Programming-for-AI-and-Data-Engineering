```python
sentence=input("Enter sentence:")
sentence=sentence.lower()
count=0

for ch in sentence:
    if ch == "a":
        count=+1
        
if count==0:
    print("No letter a")
elif count<5:
    print("Few a letters")
else:
    print("Many a letters")
```

```python
password=input("Enter password")

digit_count=0
has_upper=False
has_special=False

for ch in password:
    if ch.isdigit():
        digit_count+=1
    elif ch.isupper():
        has_upper=True
    elif not ch.isalnum():
        has_special=True

if len(password)<8:
    print("Weak password")
elif digit_count==0 or not has_upper:
    print("Medium password")
elif digit_count>0 and has_upper and has_special:
    print("Strong password")
else:
    print("Medium password")
```

```python
correct_password="python123"
attemps=3

while attemps>0:
    password=input("Enter the password:")
    if correct_password==password:
        print("Login successfull")
        break
    else:
        attemps-=1
        if attemps>0:
            print(f"Remaining attemps {attemps}")

if attemps==0:
    print("Account locked")
```
