```python
def add(a, b):
    return a + b

def subtract(a, b):
    return a - b

def multiply(a, b):
    return a * b

def divide(a, b):
    if b == 0:
        return "Error: Cannot divide by zero"
    return a / b

operator = input("Enter an operator (+, -, *, /): ")
num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

if operator == '+':
    print("Result:", add(num1, num2))
elif operator == '-':
    print("Result:", subtract(num1, num2))
elif operator == '*':
    print("Result:", multiply(num1, num2))
elif operator == '/':
    print("Result:", divide(num1, num2))
else:
    print("Invalid operator entered.")
```

```python
def main():
    my_list = [2, 4, 6, 10]
    analyze_squared_list(my_list)

def analyze_squared_list(my_list):
    if not my_list:
        print("List is empty!")
        return
    
    squared_values = []
    for x in my_list:
        square = x ** 2
        squared_values.append(square)
        
    max_val = max(squared_values)
    min_val = min(squared_values)
    
    print("Original list:", my_list)
    print("Squared values:", squared_values)
    print("Maximum squared value:", max_val)
    print("Minimum squared value:", min_val)

main()
```
