```python
products=[]
prices=[]

for i in range(5):
    name=input(f"Enter {i+1}. product name: ")
    price=float(input(f"Enter price of {name} :"))

    products.append(name)
    prices.append(price)

total=sum(prices)
max_price=max(prices)
min_price=min(prices)

max_product=products[prices.index(max_price)]
min_product=products[prices.index(min_price)]

print(f"Total price: ", total)
print(f"Most expensive product: ", max_product)
print(f"Cheapest product: ", min_product)

if total<500:
    print("Small purchase")
elif total<2000:
    print("Medium purchase")
else:
    print("Large purchase")
```

```python
sentence=input("Enter a sentence:")
words=sentence.split()
longest=words[0]
shortest=words[0]

for word in words:
    if len(word)>len(longest):
        longest=word
    elif len(word)<len(shortest):
        shortest=word
lenght=len(longest)
if lenght<5:
    print("short words")
elif lenght<=8:
    print("medium words")
else:
    print("long words")
```

```python
nums=[]
while True:
    print("1- Add number")
    print("2- Show number")
    print("3- Show average")
    print("4- Show maximum")
    print("0- Exit")

    choose=input("Select a choose:")

    if choose=="1":
        number=int(input("Enter a number:"))
        nums.append(number)
    elif choose=="2":
        print(nums)
    elif choose=="3":
        if len(nums)==0:
            print("No number in list")
        else:
            avg=sum(nums)/len(nums)
            print(f"average: ", avg)
    elif choose=="4":
        if len(nums)==0:
            print("No number in list")
        else:
            print(f"Maximum number of list: {max(nums)}")
    elif choose=="0":
        break
    else:
        print("Invalid entered")
```

```python
orders=[]
for i in range(7):
    order=input("Enter order name:")
    orders.append(order)

print(orders)
product=input("Enter search product name:")
if product in orders:
    count=orders.count(product)
    print(f"{product} {count} times ordered.")

    if count==1:
        print("Rare ordered")
    elif count<=3:
        print("Popular order")
    else:
        print("Very Popular order")
else:
    print("Product not found")
```
