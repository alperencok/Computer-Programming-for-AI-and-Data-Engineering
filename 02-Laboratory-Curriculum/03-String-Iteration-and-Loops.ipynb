```python
# For Loop: Iterating over a string
password = "Password123!"
digit_count = 0

for ch in password:
    if ch.isdigit():
        digit_count += 1
print(f"There are {digit_count} digits in the password.")

# For Loop: Using range()
for i in range(1, 4):  # Loops through 1, 2, 3
    print(f"Student {i}")

# While Loop: Runs as long as the condition is True
attempts = 3
while attempts > 0:
    print(f"You have {attempts} attempts left.")
    attempts -= 1  # Decrease attempt by 1
print("Out of attempts!")
```

```python
sentence = input("Enter a sentence: ")
sentence = sentence.lower()
count = 0

for ch in sentence:
    if ch == "a":
        count += 1
        
if count == 0:
    print("No letter a")
elif count < 5:
    print("Few a letters")
else:
    print("Many a letters")
```

```python
password = input("Enter password: ")

digit_count = 0
has_upper = False
has_special = False

for ch in password:
    if ch.isdigit():
        digit_count += 1
    elif ch.isupper():
        has_upper = True
    elif not ch.isalnum():
        has_special = True

if len(password) < 8:
    print("Weak password")
elif digit_count == 0 or not has_upper:
    print("Medium password")
elif digit_count > 0 and has_upper and has_special:
    print("Strong password")
else:
    print("Medium password")
```

```python
correct_answer = "ABCDE"

for i in range(1, 4):
    answer = input(f"Student {i}: ")
    grade = 0
    for j in range(len(correct_answer)):
        if correct_answer[j] == answer[j]:
            grade += 20
    print(f"Student {i} Grade: {grade}")
```

```python
correct_password = "python123"
attempts = 3

while attempts > 0:
    password = input("Enter the password: ")
    if correct_password == password:
        print("Login successful")
        break
    else:
        attempts -= 1
        if attempts > 0:
            print(f"Remaining attempts: {attempts}")

if attempts == 0:
    print("Account locked")
```
