## 1. 📊 Series

A `Series` is a one-dimensional labeled data structure. Think of it like a list, but every element has an index (label).

```python
# Creating a Series from a list
s1 = pd.Series([10, 20, 30, 40, 50])
print("Series from list:")
print(s1)
print(f"\nType: {s1.dtype}, Shape: {s1.shape}")
```

```python
# Series with custom index
s2 = pd.Series([23.5, 18.0, 31.2, 27.8],
               index=['January', 'February', 'March', 'April'],
               name='Average Temperature (°C)')
print(s2)
print(f"\nMarch temperature: {s2['March']}°C")
```

```python
# Series from a dictionary
s3 = pd.Series({'Apple': 5, 'Banana': 3, 'Orange': 8, 'Strawberry': 12})
print("Series from dictionary:")
print(s3)

# Mathematical operation on a Series
print("\nDoubled:")
print(s3 * 2)
```

```python
(s3*2)['Apple']
s3['Apple']
```

## 2. 📋 Creating DataFrames

A `DataFrame` is a two-dimensional labeled data structure. Think of it like an Excel spreadsheet.

```python
# DataFrame from a dictionary
df_student = pd.DataFrame({
    'Name': ['Ali', 'Ayse', 'Mehmet', 'Fatma', 'Can'],
    'Age': [20, 22, 21, 23, 19],
    'Grade': [85, 92, 78, 95, 88],
    'City': ['Istanbul', 'Ankara', 'Izmir', 'Istanbul', 'Bursa']
})
print("Student DataFrame:")
df_student
```

## 3. 🔍 Exploring Data

Essential methods for understanding your data.

```python
# Build a sample dataset
np.random.seed(0)
df = pd.DataFrame({
    'name': ['Ali', 'Ayse', 'Mehmet', 'Fatma', 'Can', 'Zeynep', 'Hasan', 'Merve'],
    'age': [25, 30, 35, 28, 22, 33, 27, 29],
    'salary': [5000, 7500, 9000, 6500, 4500, 8000, 5500, 7000],
    'department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT'],
    'experience': [2, 5, 10, 4, 1, 8, 3, 6]
})

print("First 5 rows (head):")
print(df.head())
```

```python
print("Last 3 rows (tail):")
print(df.tail(3))
```

## 4. 🎯 Data Selection & Indexing

Methods for accessing specific rows and columns.

```python
# Single column selection
print("Names (single column):")
print(df['name'])
```

```python
# Multiple column selection
print("Name and Salary:")
print(df[['name', 'salary']])
```

```python
# .loc - Label-based selection
print(".loc row selection (index 2):")
print(df.loc[2])

print("\n.loc row and column selection:")
print(df.loc[1:3, ['name', 'salary', 'department']])
```

```python
# .iloc - Position-based selection
print(".iloc first 3 rows, first 2 columns:")
print(df.iloc[:3, :2])

print("\n.iloc specific cell (row 0, column 2):")
print(df.iloc[0, 2])
```

```python
# at and iat - fast single-cell access
print(f".at access (index 3, 'name'): {df.at[3, 'name']}")
print(f".iat access (row 3, column 0): {df.iat[3, 0]}")
```

## 5. 🔎 Data Filtering

Filtering data based on conditions.

```python
# Simple condition
print("Employees with salary > 7000:")
print(df[df['salary'] > 7000])
```

```python
# Multiple conditions (&, |)
print("IT department AND salary > 5000:")
mask = (df['department'] == 'IT') & (df['salary'] > 5000)
print(df[mask])
```

```python
# isin() - filter by a list of values
print("IT or Finance department:")
print(df[df['department'].isin(['IT', 'Finance'])])
```

```python
# str.contains() - filter rows containing text
print("Names containing 'a' or 'A':")
print(df[df['name'].str.contains('a', case=False)])
```

```python
# between() - range filter
print("Age between 25 and 30:")
print(df[df['age'].between(25, 30)])
```

## 6. 🧹 Data Cleaning

Handling missing values, duplicates, and reshaping data.

```python
# DataFrame with missing values
df_missing = pd.DataFrame({
    'name': ['Alice', 'Bob', None, 'Diana', 'Eve'],
    'age': [25, None, 35, 28, None],
    'score': [85.0, 92.0, None, 95.0, 88.0]
})
print("DataFrame with missing values:")
print(df_missing)
print("\nMissing value count:")
print(df_missing.isnull().sum())
```

```python
# isna() and notna()
print("Which cells are empty?")
print(df_missing.isna())
```

```python
# dropna() - remove rows with missing values
print("Drop all missing rows (dropna):")
print(df_missing.dropna())

print("\nDrop rows missing only in 'name':")
print(df_missing.dropna(subset=['name']))
```

```python
# fillna() - fill missing values
print("Fill numeric missing values with mean/median:")
df_filled = df_missing.copy()
df_filled['age'] = df_filled['age'].fillna(df_filled['age'].mean())
df_filled['score'] = df_filled['score'].fillna(df_filled['score'].median())
df_filled['name'] = df_filled['name'].fillna('Unknown')
print(df_filled)
```

```python
# Duplicates
df_dup = pd.DataFrame({
    'name': ['Alice', 'Bob', 'Alice', 'Charlie', 'Bob'],
    'score': [85, 90, 85, 75, 90]
})
print("Original:")
print(df_dup)
print("\nDuplicate rows:")
print(df_dup.duplicated())
print("\nAfter removing duplicates:")
print(df_dup.drop_duplicates())
```

```python
# rename() - rename columns
df_renamed = df[['name', 'salary', 'age']].copy()
df_renamed = df_renamed.rename(columns={'name': 'full_name', 'salary': 'annual_salary', 'age': 'years_old'})
print("Renamed columns:")
print(df_renamed.head())
```

## 7. 🔄 Data Transformation

Processing and transforming data.

```python
df2 = df.copy()

def classify_salary(salary):
    if salary < 5500:
        return 'Low'
    elif salary < 7500:
        return 'Medium'
    else:
        return 'High'

df2['salary_level'] = df2['salary'].apply(classify_salary)
print("Salary level classification:")
print(df2[['name', 'salary', 'salary_level']])
```

```python
# map() - map values in a Series
dept_map = {'IT': 'Information Technology', 'HR': 'Human Resources', 'Finance': 'Financial Affairs'}
df2['department_full'] = df2['department'].map(dept_map)
print("Full department names (map):")
print(df2[['name', 'department', 'department_full']])
```

```python
# sort_values() - sorting
print("Sort by salary descending:")
print(df.sort_values('salary', ascending=False))

print("\nSort by department ascending, salary descending:")
print(df.sort_values(['department', 'salary'], ascending=[True, False]))
```

```python
# astype() - convert data types
print(f"Before - age dtype: {df['age'].dtype}")
df3 = df.copy()
df3['age'] = df3['age'].astype(float)
print(f"After  - age dtype: {df3['age'].dtype}")

# Categorical type
df3['department'] = df3['department'].astype('category')
print(f"Categories: {df3['department'].cat.categories.tolist()}")
```

```python
df3
```

```python
# cut() and qcut() - bin numeric data into groups
df3['age_group'] = pd.cut(df3['age'],
                          bins=[0, 25, 30, 100],
                          labels=['Young (≤25)', 'Mid (26-30)', 'Senior (>30)'])
print("Age groups (cut):")
print(df3[['name', 'age', 'age_group']])
```

## 8. 📦 Grouping & Aggregation

Analyzing data by categories using `groupby()`.

```python
# Basic groupby
print("Average salary by department:")
print(df.groupby('department')['salary'].median())
```

```python
# Multiple aggregations - agg()
print("Salary statistics by department:")
print(df.groupby('department')['salary'].agg(['min', 'max', 'mean', 'count']))
```

```python
# Different aggregations for multiple columns
print("Detailed aggregation:")
result = df.groupby('department').agg(
    avg_salary=('salary', 'mean'),
    total_salary=('salary', 'sum'),
    headcount=('name', 'count'),
    avg_experience=('experience', 'mean')
).round(1)
print(result)
```

```python
# transform() - add group-level value back to original DataFrame
df4 = df.copy()
df4['dept_avg_salary'] = df4.groupby('department')['salary'].transform('mean')
df4['salary_diff'] = df4['salary'] - df4['dept_avg_salary']
print("Salary difference from department average:")
print(df4[['name', 'department', 'salary', 'dept_avg_salary', 'salary_diff']].round(1))
```

## 9. 🔗 Joining & Merging

Combining DataFrames.

```python
# Sample DataFrames
df_employees = pd.DataFrame({
    'id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'Diana', 'Eve'],
    'dept_id': [10, 20, 10, 30, 50]
})

df_departments = pd.DataFrame({
    'dept_id': [10, 20, 30, 40],
    'dept_name': ['IT', 'HR', 'Finance', 'Marketing'],
    'location': ['New York', 'Chicago', 'Boston', 'Seattle']
})

print("Employees:")
print(df_employees)
print("\nDepartments:")
print(df_departments)
```

```python
# merge() - SQL-style JOIN
print("Inner Join (default):")
print(pd.merge(df_employees, df_departments, on='dept_id'))

print("\nLeft Join (all employees):")
print(pd.merge(df_employees, df_departments, on='dept_id', how='left'))

print("\nRight Join (all departments):")
print(pd.merge(df_employees, df_departments, on='dept_id', how='right'))
```

```python
# concat() - stack DataFrames vertically or horizontally
df_a = pd.DataFrame({'name': ['Alice', 'Bob'], 'score': [85, 90]})
df_b = pd.DataFrame({'name': ['Charlie', 'Diana'], 'score': [78, 95]})
print(df_a)
print(df_b)
print("Vertical concat:")
print(pd.concat([df_a, df_b], ignore_index=True))
```

## 12. ✍️ String Operations

Working with text data using the `.str` accessor.

```python
names = pd.Series(['  alice smith  ', 'BOB JONES', 'charlie brown', 'DiAnA PrInCe'])

print("Original:")
print(names.tolist())

print("\nUppercase (upper):")
print(names.str.upper().tolist())

print("\nLowercase (lower):")
print(names.str.lower().tolist())

print("\nTitle case (title):")
print(names.str.strip().str.title().tolist())

print("\nStrip whitespace (strip):")
print(names.str.strip().tolist())
```

```python
# split, replace, len
full_names = pd.Series(['Alice Smith', 'Bob Jones', 'Charlie Brown'])

print("Split first and last name:")
split_names = full_names.str.split(' ', expand=True)
split_names.columns = ['first_name', 'last_name']
print(split_names)

print("\nCharacter count:")
print(full_names.str.len().tolist())
```

# ✅ ANSWERS

### ✅ Answer 1 - Missing Value Analysis

```python
print("Missing values per column:")
print(df_market.isnull().sum())

print(f"\nTotal missing values: {df_market.isnull().sum().sum()}")

print("\nMissing value percentage:")
print((df_market.isnull().sum() / len(df_market) * 100).round(1))
```

### ✅ Answer 2 - Filling Missing Values

```python
df_clean = df_market.copy()

# quantity -> fill with mean
mean_qty = df_clean['quantity'].mean()
df_clean['quantity'] = df_clean['quantity'].fillna(mean_qty)

# unit_price -> fill with median
median_price = df_clean['unit_price'].median()
df_clean['unit_price'] = df_clean['unit_price'].fillna(median_price)

# Recalculate total_price
df_clean['total_price'] = df_clean['quantity'] * df_clean['unit_price']

print(f"Missing values after filling: {df_clean.isnull().sum().sum()}")
print(f"Mean quantity: {mean_qty:.2f}, Median unit price: {median_price:.2f}")
print("\nFirst 10 rows:")
print(df_clean.head(10))
```

### ✅ Answer 3 - Customer Total Spending

```python
customer_spending = (
    df_clean
    .groupby('customer')['total_price']
    .sum()
    .sort_values(ascending=False)
    .round(2)
)
print("Customer total spending (highest to lowest):")
print(customer_spending.to_string())
print(f"\nTop spender: {customer_spending.idxmax()} (${customer_spending.max():.2f})")
```

### ✅ Answer 4 - Product Analysis

```python
product_analysis = df_clean.groupby('product').agg(
    total_quantity=('quantity', 'sum'),
    avg_unit_price=('unit_price', 'mean'),
    total_revenue=('total_price', 'sum')
).round(2)

print("Product analysis:")
print(product_analysis)
print(f"\nBest-selling product: {product_analysis['total_quantity'].idxmax()}")
print(f"Highest revenue product: {product_analysis['total_revenue'].idxmax()}")
```

### ✅ Answer 5 - Filtering

```python
high_orders = df_clean[df_clean['total_price'] > 100]

print(f"Orders over $100: {len(high_orders)}")
print("\nThese orders:")
print(high_orders[['order_id', 'customer', 'product', 'quantity', 'total_price']].to_string(index=False))

print("\nCustomer breakdown:")
print(high_orders['customer'].value_counts())
```

### ✅ Answer 6 - Discount Column

```python
def calculate_discount(total):
    if total > 100:
        return 0.10  # 10%
    elif total > 50:
        return 0.05  # 5%
    else:
        return 0.00  # 0%

df_clean['discount'] = df_clean['total_price'].apply(calculate_discount)
df_clean['discounted_price'] = df_clean['total_price'] * (1 - df_clean['discount'])

print("Discounts applied:")
print(df_clean[['order_id', 'customer', 'total_price', 'discount', 'discounted_price']].head(15).to_string(index=False))

print("\nDiscount distribution:")
print(df_clean['discount'].value_counts().sort_index())
```

### ✅ Answer 7 - Monthly Revenue

```python
df_clean['month'] = df_clean['date'].dt.month
df_clean['month_name'] = df_clean['date'].dt.strftime('%B')

monthly_revenue = (
    df_clean.groupby(['month', 'month_name'])['total_price']
    .sum()
    .round(2)
    .reset_index()
    .sort_values('month')
)

print("Monthly total revenue:")
print(monthly_revenue[['month_name', 'total_price']].to_string(index=False))

best_idx = monthly_revenue['total_price'].idxmax()
best_month = monthly_revenue.loc[best_idx]
print(f"\nBest month: {best_month['month_name']} (${best_month['total_price']:.2f})")
```

### ✅ Answer 8 - String Operations

```python
df_clean['customer_code'] = df_clean['customer'].str.upper() + '_CUSTOMER'

print("Customer codes:")
print(df_clean[['customer', 'customer_code']].drop_duplicates().reset_index(drop=True))
```
