Q1) Write the Python function that checks whether the given number is a strong number.<br/>
145 = 1! + 4! + 5!

```python
def factorial(number):
    fact=1
    for i in range(1,number+1):
        fact=fact*i
    return fact
def is_strong_number(number):
    total=0
    temp=number
    while number>0:
        total=total+factorial(number%10)
        number=number//10
    return total==temp

def is_strong_number2(number):
    return number==sum(factorial(int(d)) for d in str(number))

print(is_strong_number(145))
print(is_strong_number2(145))
```

Q2) Write a function that determines whether or not a password is good. We will define a good password to be a one that is at least 8 characters long and contains at least one uppercase letter, at least one lowercase letter, and at least one number.

```python
def is_good_password(password):
    if len(password)<8:
        return False
    isupper, islower,isdigit=False,False,False
    for c in password:
        if c.isupper():
            isupper=True
        elif c.islower():
            islower=True 
        elif c.isdigit():
            isdigit=True 
    return isupper and islower and isdigit

is_good_password("example1E")
```

Q3) A number is called a non-palindromic prime number if it is prime and its reverse is also prime and the number is not palindromic.

For example: 13 is prime and 31 is prime, so 13 and 31 are pairs of non-palindromic primes.

Write a Python code that prints the non-palindromic primes between 1 and 400.
The signatures of the methods you will use should be as follows:
- get_reverse(number)
- is_palindrome(number)
- is_prime(number)
- print_non_palindromic_primes()

```python
def get_reverse(number):
    return int(str(number)[::-1])
def is_palindrome(number): # 1221- 1221
    return number==get_reverse(number)
def is_prime(number):
    if number<2:
        return False
    for i in range(2,number):
        if number%i==0:
            return False
    return True
def print_non_palindromic_primes():
    for i in range(1,400):
        if is_prime(i) and is_prime(get_reverse(i)) and not is_palindrome(i):
            print(i)   
print_non_palindromic_primes()
```
