```python
# Basic conditional logic
age = int(input("Enter your age: "))

if age < 18:
    print("You are a minor.")
elif age >= 18 and age < 65:
    print("You are an adult.")
else:
    print("You are a senior citizen.")
```

```python
sentence = "I am learning Python programming"
keyword = "python"

# Check if a substring exists within a string (case-insensitive)
if keyword.lower() in sentence.lower():
    print("Keyword found!")
    # Replace keyword and convert to uppercase
    sentence = sentence.replace(keyword, keyword.upper())
    print("Updated sentence:", sentence)
else:
    print("Keyword not found!")
```

```python
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter an operation (+, -, *, /): ")

if operation == "+":
    print("Result:", num1 + num2)
elif operation == "-":
    print("Result:", num1 - num2)
elif operation == "*":
    print("Result:", num1 * num2)
elif operation == "/":
    if num2 != 0:
        print("Result:", num1 / num2)
    else:
        print("Error: Cannot divide by zero!")
else:
    print("Invalid operation")
```

```python
email = input("Enter email: ").lower()
birth_year = int(input("Enter birth year: "))

# Basic email validation
if "@" not in email or "." not in email:
    print("Invalid email.")
else:
    domain = email.split("@")[1]

    if domain == "edu.tr":
        age = 2026 - birth_year
        if age > 18:
            print("Student access denied.")
        else:
            print("Student access granted.")
    else:
        print("Normal user.")
```

```python
product_name = input("Enter product name: ")
category = input("Enter category (electronics, clothing, food): ").lower()
price = float(input("Enter price: "))
quantity = int(input("Enter quantity: "))
coupon_code = input("Enter coupon code: ").lower()

total_price = price * quantity
discount = 0

if category == "electronics" and total_price > 10000:
    discount = 10
elif category == "clothing" and "sale" in coupon_code:
    discount = 20
elif category == "food":
    discount = 5

discount_amount = total_price * (discount / 100)
final_price = total_price - discount_amount

print("\n--- INVOICE ---")
print("PRODUCT:", product_name.upper())
print("Category:", category)
print(f"Total Price: {total_price:.2f} TL")
print(f"Discount: %{discount}")
print(f"Final Price: {final_price:.2f} TL")
```

```python
member_name = input("Enter member name: ")
member_type = input("Enter member type (student, regular, premium): ").lower()
book_type = input("Enter book type (novel, textbook, magazine): ").lower()
borrow_days = int(input("Enter late days: "))

daily_penalty = 0

if member_type == "student":
    if book_type == "novel": daily_penalty = 1
    elif book_type == "textbook": daily_penalty = 2
    elif book_type == "magazine": daily_penalty = 0.5
elif member_type == "regular":
    if book_type == "novel": daily_penalty = 2
    elif book_type == "textbook": daily_penalty = 3
    elif book_type == "magazine": daily_penalty = 1
elif member_type == "premium":
    daily_penalty = 0

total_penalty = daily_penalty * borrow_days
status = "High penalty" if total_penalty > 50 else "Normal penalty"

print("\n--- PENALTY REPORT ---")
print("MEMBER:", member_name.upper())
print("Member Type:", member_type)
print("Book Type:", book_type)
print("Late Days:", borrow_days)
print(f"Total Penalty: {total_penalty:.2f} TL")
print("Status:", status)
```
