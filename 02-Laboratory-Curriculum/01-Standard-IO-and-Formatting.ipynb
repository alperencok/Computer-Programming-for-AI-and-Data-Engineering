```python
#1. Print Function and Escape Characters
```

```python
# sep (separator) and end parameters
print("apple", "banana", "orange", sep=", ")
print("2024", "12", "25", sep="-")
print("Hello", "World", end="---\n")

# Escape characters and Raw String
print("Backslash: \\")
print('Single quotes: \'hello\'')
print("Double quotes: \"hello\"")

# \n moves to a new line, 'r' (raw) ignores escape characters
print('C:\\some\\name')   
print(r'C:\some\name')    

# String concatenation and multiplication
message = 'Hello ' + 'World'
print(message)
print(3 * 'un' + 'ium')
```

```python
#2. Mathematical Operations and Float Precision
```

```python
# Basic Operators
print("Addition:", 5 + 10)
print("Subtraction:", 5 - 10)
print("Multiplication:", 4 * 8)
print("Division (Returns Float):", 17 / 3)
print("Floor Division (Returns Integer):", 17 // 3)
print("Modulus/Remainder:", 17 % 3)
print("Exponentiation:", 2 ** 7)       # 128
print("Negative Exponent:", 2 ** -2)   # 0.25 (1/4)

# Order of Operations
print("5 + (3 * 2):", 5 + 3 * 2) 

# Float Precision Issue and Solution (Decimal)
# Note: 0.1 + 0.2 does not exactly equal 0.3 in Python due to base-2 representation.
print("0.1 + 0.2 == 0.3 ->", 0.1 + 0.2 == 0.3) # Returns False

from decimal import Decimal
a = Decimal('0.1')
b = Decimal('0.2')
c = Decimal('0.3')
print("Using Decimal: 0.1 + 0.2 == 0.3 ->", c == (a + b)) # Returns True
```

```python
#3. Variables, Data Types, and Input
```

```python
course_name = "Python Programming"
number_of_students = 30

# Checking data types
print(type(course_name))  # <class 'str'>
print(type(number_of_students)) # <class 'int'>

# Getting input from the user and Type Casting
x_str = input("Enter a number: ")
y_int = int(x_str)

# Difference between String concatenation and Integer addition
print("As String (Concatenation):", x_str + "1")
print("As Integer (Addition):", y_int + 1)
```

```python
#4. Strings (Indexing, Slicing, and Methods)
```

```python
word = 'Python'

# Indexing and Slicing
print(word[0])    # First character: P
print(word[-1])   # Last character: n
print(word[0:2])  # First two characters: Py
print(word[2:])   # From the 2nd index to the end: thon
print(word[::-1]) # Reverse the string: nohtyP

# String Methods
text = "   Python Course   "
print("Upper:", text.upper())
print("Lower:", text.lower())
print("Strip (Removes leading/trailing spaces):", text.strip())
print("Lstrip (Removes leading spaces only):", text.lstrip())
print("Rstrip (Removes trailing spaces only):", text.rstrip())

# String replacement (Strings are immutable, so we overwrite the variable)
word = 'J' + word[1:]
print("Modified string:", word) # Jython
```

```python
#5. String Formatting (F-Strings)
```

```python
num = 12345.6789
discount = 0.5

print(f"Original number: {num}")
print(f"Rounded (2 decimals): {num:.2f}")
print(f"With thousand separator: {num:,.2f}")
print(f"Scientific (Exponential) notation: {num:.2e}")
print(f"Percentage format: {discount:.0%}")

# String Alignment
print(f"Left aligned  : |{num:<20.2f}|")
print(f"Right aligned : |{num:>20.2f}|")
print(f"Center aligned: |{num:^20.2f}|")
```

```python
#6. Boolean Operators and Comparisons
```

```python
# Comparisons (Returns True/False)
print("5 > -2 ?", 5 > -2)
print("5 <= -2 ?", 5 <= -2)
print("3 + 2 < 5 - 7 ?", 3 + 2 < 5 - 7)

# Logical Operators (and, or, not)
print("True and False:", True and False)
print("True or False:", True or False)
print("True and not False:", True and not False)

# None Value Check
a = None
print("Is 'a' not None?", a is not None)
```

```python
# --- Exercise 1: Mathematical Formula ---
# Calculate: (x - 3)^2 + (x + 1)^2
x = float(input("Enter a number: "))
result = (x - 3)**2 + (x + 1)**2

print(f"Entered number: {x}")
print(f"Calculated result: {result:.2f}")


# --- Exercise 2: Student Grade Average ---
student_name = input("Student name: ")
course_name = input("Course name: ")
score1 = float(input("Score 1: "))
score2 = float(input("Score 2: "))
score3 = float(input("Score 3: "))

average = (score1 + score2 + score3) / 3

print("\n--- Result ---")
print(f"Student: {student_name}")
print(f"Course: {course_name}")
print(f"Average: {average:.2f}")


# --- Exercise 3: Salary, Expenses, and Savings Calculation ---
name = input("Name: ")
birth_year = int(input("Birth year: "))
salary = float(input("Monthly salary: "))
expenses = float(input("Monthly expenses: "))

age = 2026 - birth_year
remaining = salary - expenses
savings = remaining * 0.15
spendable = remaining - savings

print(f"\n{name} ({age} years old)")
print(f"Salary: {salary:,.2f} TL")
print(f"Expenses: {expenses:,.2f} TL")
print(f"Remaining: {remaining:,.2f} TL")
print(f"Savings (15%): {savings:,.2f} TL")
print(f"Spendable: {spendable:,.2f} TL")


# --- Exercise 4: String Cleaning and Password Masking ---
username = input("Username: ")
email = input("Email: ")
password = input("Password: ")

# 1. Remove spaces and convert to lowercase
clean_username = username.strip().lower()

# 2. Convert email to lowercase and split it
clean_email = email.lower()
email_user, email_domain = clean_email.split("@")

# 3. Password length check
if len(password) < 8:
    print("\nWarning: Password is too short!")

# 4. Mask the password (First letter + stars + last letter)
if len(password) >= 2:
    masked_password = password[0] + ("*" * (len(password) - 2)) + password[-1]
else:
    masked_password = password

# Print results
print("\n--- User Profile ---")
print(f"Cleaned Username: {clean_username}")
print(f"Email Username: {email_user}")
print(f"Email Domain: {email_domain}")
print(f"Masked Password: {masked_password}")
```
